```python
firsMas = np.array([10, 3, 1])
secondMas = np.array([2, 5, 8])

vertical = np.vstack((firsMas, secondMas))
horizontal = np.hstack((firsMas, secondMas))
print("Vertical combination:", vertical)
print("Horizontal combination:", horizontal)
```

7. Визначити функцію, яка приймає на вході масив і транспонує його. Скористайтесь методом reshape. І не забудьте почати з docstrings.

```python
def transponMas(mas):
  """
    Functoin for transposing an masive
  """
  return mas.reshape(mas.shape[::-1]);

array = np.array([[2, 4, 6], [5, 3, 1]])
transponMas = transponMas(array)
print(f"Transposed array from: \n {array}, \n to: \n {transponMas}")
```

8. Створіть два масиви. Використовучи ці масиви, продемонструйте 1) операцію по-елементного додавання, 2) операцію по-елементного віднімання, 3) множення масиву на число, 4) операцію по-елементного множення, і 5) матричного множення.

```python
adding = firsMas + secondMas;
subtracting = firsMas - secondMas;
multiplicating = firsMas * 10;
elementwise_multiplicating = firsMas * secondMas;
matrix_multiplicating = firsMas.T @ secondMas;

print("The element-by-element addition operation: ", adding);
print("The element-by-element subtracting operation: ", subtracting);
print("Multiplicating operation: ", multiplicating);
print("Elementwise multiplicating operation: ", elementwise_multiplicating);
print("Matrix multiplicating operation: ", matrix_multiplicating);
```
